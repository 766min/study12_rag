```python
campus_rules = (
    "제1조 (목적)\n본 규정은 장학금 지급 기준을 정의합니다.\n\n"
    "제2조 (선발 기준)\n직전 학기 평점 3.5 이상이어야 하며 F학점이 없어야 합니다."
)
```

```python
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter

text_splitter = CharacterTextSplitter(
    chunk_size=30, # 청크 최대 글자수
    chunk_overlap=10, # 겹침 글자수
    # separators="" # 기본값 \n\n - 문단 단위 청킹
)

text_chunks = text_splitter.split_text(campus_rules)

for i, chunk in enumerate(text_chunks, 1):
    print(f"[{i} 번째] - {chunk}")
```

```python
# separactors=["\n\n", "\\n", " ", ""]  - 기본값 : 문단 단위 -> 다음 줄 -> 공백 -> 마침표
rec_splitter = RecursiveCharacterTextSplitter(
    chunk_size=30,
    chunk_overlap=10,  
)

rec_chunks = rec_splitter.split_text(campus_rules)

for i, chunk in enumerate(rec_chunks, 1):
    print(f"[{i} 번째] - {chunk}")
```

```python
raw_text = """
# 1장 2강 : 생성형 AI의 동작 원리

## 1. 토큰(Token)과 임베딩(Embedding)

### 1.1 토큰화(Tokenization)

AI는 텍스트를 그대로 읽지 않고, '토큰'이라는 최소 단위로 잘게 쪼개서 처리합니다. 영어는 단어 하나가 토큰 하나에 가깝지만, 한국어는 조사와 어미 때문에 한 단어가 여러 토큰으로 나뉘는 경우가 많습니다.

### 1.2 임베딩(Embedding)

쪼개진 토큰은 컴퓨터가 계산할 수 있도록 숫자로 이루어진 벡터로 변환됩니다. 이 과정을 임베딩이라고 부릅니다. 예를 들어 '왕'과 '여왕'은 성별을 제외하면 권력이나 지위 측면에서 매우 비슷한 위치에 놓이고, '사과'와 '바나나'는 과일이라는 의미 공간에서 서로 가까운 거리에 놓이지만, '사과'와 '우주선'은 전혀 다른 맥락이기 때문에 임베딩 공간에서 아주 멀리 떨어진 위치에 놓이게 됩니다. 이렇게 되면 AI는 단어의 겉모습이 아니라 의미적으로 유사한 단어를 수학적 거리로 계산해서 찾아낼 수 있습니다. 실제 모델은 단어 하나당 수백에서 수천 개의 차원을 사용해서 훨씬 정교하게 의미를 표현합니다.

## 2. 확률 기반 다음 단어 예측(Next-Token Prediction)

### 2.1 추론(Inference) 과정

AI는 지금까지 입력된 문맥을 바탕으로 다음에 올 가장 확률이 높은 토큰 하나를 예측하고, 그 토큰을 문장에 이어 붙인 뒤 다시 전체를 입력해서 그다음 토큰을 예측하는 과정을 반복합니다.

### 2.2 온도(Temperature)

온도 값을 낮추면 AI는 항상 확률이 가장 높은 토큰만 선택해서 일관되고 딱딱한 문장을 만들고, 온도 값을 높이면 확률이 낮은 토큰도 선택될 수 있어서 더 다양하고 창의적인 문장이 만들어집니다.
"""
```

```python
rec_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=100,
    separators=["##", "###"]
)

chunks = rec_splitter.split_text(raw_text)
for i, chunk in enumerate(chunks, 1):
    print(f"[{i} 번째] {chunk}")
    print("-" * 50)

# print(len(chunks))
# for i, chunk in enumerate(chunks, 1):
#     print(i, len(chunk), chunk[:30])
```
